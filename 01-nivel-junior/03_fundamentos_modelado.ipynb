## 1. ¿Qué es un Modelo de Datos?

### Definición

> **Modelo de Datos:** Una representación abstracta y estructurada de los datos, sus relaciones, restricciones y reglas de negocio que permite comunicar, diseñar e implementar soluciones de datos.

### ¿Por qué modelar datos?

- 📋 **Comunicación:** Lenguaje común entre negocio y TI
- 🎯 **Claridad:** Visualiza estructuras complejas
- 🔒 **Integridad:** Define reglas y restricciones
- 📐 **Diseño:** Guía la implementación técnica
- 📖 **Documentación:** Registro del conocimiento del negocio

## 2. Niveles de Abstracción en Modelado de Datos

Existen **tres niveles** principales de modelado:

### 2.1 Modelo Conceptual

**Audiencia:** Stakeholders de negocio

**Características:**
- Alto nivel de abstracción
- Independiente de la tecnología
- Enfocado en entidades y relaciones del negocio
- No incluye detalles técnicos

**Ejemplo:** "Un Cliente puede tener múltiples Pedidos"

### 2.2 Modelo Lógico

**Audiencia:** Arquitectos y analistas

**Características:**
- Nivel medio de abstracción
- Independiente del DBMS específico
- Define atributos y tipos de datos
- Especifica llaves primarias y foráneas
- Aplica normalización

**Ejemplo:** 
```
Cliente (cliente_id PK, nombre, email, fecha_registro)
Pedido (pedido_id PK, cliente_id FK, fecha, total)
```

### 2.3 Modelo Físico

**Audiencia:** DBAs e ingenieros

**Características:**
- Bajo nivel de abstracción
- Específico del DBMS (PostgreSQL, MySQL, etc.)
- Incluye índices, particiones, tablespaces
- Define almacenamiento y optimizaciones

**Ejemplo:** 
```sql
CREATE TABLE cliente (
    cliente_id INT PRIMARY KEY AUTO_INCREMENT,
    nombre VARCHAR(100) NOT NULL,
    email VARCHAR(100) UNIQUE,
    fecha_registro TIMESTAMP DEFAULT CURRENT_TIMESTAMP,
    INDEX idx_email (email)
);
```

## 4. Calidad del Dato

### Definición

> **Calidad del Dato:** El grado en que los datos son aptos para su uso previsto en operaciones, toma de decisiones y planificación.

### Dimensiones de la Calidad del Dato

#### 4.1 Exactitud (Accuracy)
- ¿Los datos reflejan correctamente la realidad?
- Ejemplo: Número de teléfono correcto

#### 4.2 Completitud (Completeness)
- ¿Están presentes todos los datos requeridos?
- Ejemplo: Todos los clientes tienen email

#### 4.3 Consistencia (Consistency)
- ¿Los datos son coherentes entre sistemas?
- Ejemplo: Mismo formato de fecha en todas las tablas

#### 4.4 Vigencia (Timeliness)
- ¿Los datos están actualizados?
- Ejemplo: Dirección actual del cliente

#### 4.5 Validez (Validity)
- ¿Los datos cumplen con reglas de negocio?
- Ejemplo: Email con formato válido

#### 4.6 Unicidad (Uniqueness)
- ¿No hay duplicados?
- Ejemplo: Un solo registro por cliente

````python
# Ejemplo de evaluación de calidad de datos
import numpy as np

# Dataset con problemas de calidad
datos_problemas = pd.DataFrame({
    'cliente_id': [1, 2, 3, 4, 5, 5],  # Duplicado
    'nombre': ['Ana García', 'Luis Pérez', None, 'Carlos Ruiz', 'Laura Martín', 'Laura Martín'],  # Valor nulo
    'email': ['ana@email.com', 'luis@invalido', 'maria@email.com', 'carlos@email.com', 'laura@email.com', 'laura@email.com'],  # Email inválido
    'edad': [25, 150, 30, -5, 28, 28],  # Edad inválida
    'fecha_registro': ['2024-01-15', '2020-05-20', '2024-03-10', '2024-04-05', None, None]  # Valor nulo
})

print("="*100)
print("ANÁLISIS DE CALIDAD DE DATOS")
print("="*100)
print("\n🔍 Dataset con problemas de calidad:")
print(datos_problemas)

# Función de evaluación de calidad
def evaluar_calidad(df):
    resultados = {}
    
    # 1. Completitud
    total_celdas = df.size
    celdas_vacias = df.isnull().sum().sum()
    completitud = ((total_celdas - celdas_vacias) / total_celdas) * 100
    resultados['Completitud'] = f"{completitud:.1f}%"
    
    # 2. Unicidad
    duplicados = df.duplicated().sum()
    unicidad = ((len(df) - duplicados) / len(df)) * 100
    resultados['Unicidad'] = f"{unicidad:.1f}%"
    
    # 3. Validez (ejemplo: edad entre 0 y 120)
    if 'edad' in df.columns:
        edades_validas = df['edad'].between(0, 120).sum()
        validez_edad = (edades_validas / len(df)) * 100
        resultados['Validez (edad)'] = f"{validez_edad:.1f}%"
    
    # 4. Validez (ejemplo: formato email)
    if 'email' in df.columns:
        emails_validos = df['email'].str.contains('@.*\.', na=False).sum()
        validez_email = (emails_validos / len(df)) * 100
        resultados['Validez (email)'] = f"{validez_email:.1f}%"
    
    return resultados

print("\n📊 Métricas de Calidad:")
print("-" * 100)
resultados = evaluar_calidad(datos_problemas)
for dimension, valor in resultados.items():
    print(f"  {dimension}: {valor}")

print("\n⚠️ Problemas Identificados:")
print("-" * 100)
print("  • Cliente ID 5 está duplicado")
print("  • Registro 3 tiene nombre nulo")
print("  • Cliente 2 tiene email sin formato válido")
print("  • Cliente 2 tiene edad imposible (150)")
print("  • Cliente 4 tiene edad negativa (-5)")
print("  • Cliente 5 tiene fecha de registro nula")
````

## 5. Principios Básicos de Modelado

### 5.1 Principios Fundamentales

1. **Simplicidad:** Modelo lo más simple posible
2. **Claridad:** Nombres descriptivos y autodocumentados
3. **Flexibilidad:** Fácil de modificar y extender
4. **Normalización apropiada:** Según el caso de uso
5. **Integridad referencial:** Relaciones bien definidas

### 5.2 Mejores Prácticas

#### Nomenclatura
- Usar nombres claros y consistentes
- snake_case para tablas y columnas
- Prefijos para indicar tipo (dim_, fact_, stg_)

#### Llaves
- Siempre definir llaves primarias
- Preferir llaves surrogadas (IDs artificiales)
- Documentar llaves foráneas

#### Tipos de Datos
- Usar el tipo más apropiado y restrictivo
- Considerar el rango de valores
- Pensar en el futuro (escalabilidad)

````python
# Ejemplo de buenas vs malas prácticas
print("="*100)
print("COMPARACIÓN: BUENAS VS MALAS PRÁCTICAS EN MODELADO")
print("="*100)

comparacion_practicas = pd.DataFrame({
    'Aspecto': [
        'Nombre de tabla',
        'Nombre de columna',
        'Tipo de dato',
        'Llave primaria',
        'Valores nulos',
        'Documentación'
    ],
    '❌ Mala Práctica': [
        'tbl1, datos, x',
        'col1, campo1, x',
        'VARCHAR(255) para todo',
        'No definida',
        'Permitidos sin justificación',
        'Sin documentar'
    ],
    '✅ Buena Práctica': [
        'clientes, dim_productos',
        'cliente_id, fecha_registro',
        'INT, DATE, DECIMAL(10,2)',
        'cliente_id PRIMARY KEY',
        'NOT NULL donde sea crítico',
        'Metadatos completos'
    ]
})

print("\n")
for idx, row in comparacion_practicas.iterrows():
    print(f"\n{row['Aspecto']}:")
    print(f"  ❌ {row['❌ Mala Práctica']}")
    print(f"  ✅ {row['✅ Buena Práctica']}")
````

## 15. Actualización 2025: Fundamentos de Modelado de Datos

### Tendencias Relevantes
- Capas semánticas y metric stores (dbt Semantic Layer, Transform, Cube) ayudan a estandarizar métricas y reducir ambigüedad.
- Data Products diseñados con contratos de datos: campos obligatorios, tipos, SLAs, reglas de calidad; impulsan interoperabilidad en arquitecturas distribuídas.
- Combinación de modelado dimensional + enfoque wide tables para casos de exploración ad-hoc y velocidad (ej. DuckDB sobre vistas materializadas). 
- Uso creciente de Columnar en analítica (Parquet, ORC) y Formatos ACID (Delta, Iceberg, Hudi) para evolución controlada del esquema.
- Documentación automatizada (DataHub, OpenMetadata) captura relaciones y linaje sin esfuerzo manual exclusivo.

### Patrones Modernos (Sin Alucinación)
| Patrón | Uso Adecuado | Herramientas/Formatos |
|--------|--------------|-----------------------|
| Dimensional (Star/Snowflake) | BI clásico, métricas conformadas | SQL, dbt, Parquet/Delta |
| Wide Table | Exploración rápida, ML feature baseline | DuckDB, BigQuery |
| Data Vault (core) | Auditoría, historización granular | Hubs, Links, Satellites |
| Entidad-Atributo-Valor (EAV) | Atributos altamente variables | Usar con cuidado (complejidad joins) |
| SCD Tipo 2 | Cambios históricos en dimensiones lentamente cambiantes | Delta/Iceberg + merge semántico |

### Checklist Granularidad
1. Definición clara del nivel del hecho (transacción / línea / evento). 
2. No mezclar granularidades en misma tabla (evitar duplicidades).
3. Dimensiones conformadas coherentes (llaves surrogate si aplica). 
4. Métricas derivadas en vistas, no duplicar lógica en múltiples capas.
5. Asegurar densidad adecuada (evitar tablas extremadamente dispersas sin justificación).
