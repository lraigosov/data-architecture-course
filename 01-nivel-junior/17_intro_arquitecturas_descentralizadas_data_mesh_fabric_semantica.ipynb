# Introducción a Arquitecturas Descentralizadas: Data Mesh, Data Fabric y Capa Semántica

**Objetivo:** Entender por qué las organizaciones migran de arquitecturas centralizadas monolíticas de datos a modelos descentralizados orientados a dominios con mayor interoperabilidad y gobernanza distribuida.

---
## 1. ¿Por qué cambiar el modelo centralizado?
Los data lakes/warehouses centralizados enfrentan cuellos de botella: backlog infinito, conocimiento tribal, lentitud para nuevos casos de uso y bajo *ownership* del dominio.

| Problema Centralizado | Impacto | Enfoque Descentralizado |
|------------------------|---------|--------------------------|
| Backlog único | Cuellos de entrega | Dominios publican productos de datos |
| Esquemas globales rígidos | Fragilidad ante cambios | Contratos por dominio |
| Data Swamp | Pérdida de confianza | Gobernanza federada + calidad automatizada |
| Shadow pipelines | Costos y riesgo | Self-service gobernado |
---
## 2. Principios Fundamentales de Data Mesh (versión simplificada)
1. Datos como Producto (valor, usabilidad, calidad, documentación).
2. Ownership Descentralizado por Dominios.
3. Plataforma Self-Service (infra abstraída).
4. Gobernanza Federada Automatizada (políticas + controles como código).

### Datos como Producto (Dimensiones)
| Dimensión | Pregunta Clave | Ejemplo Métrica |
|-----------|-----------------|-----------------|
| Descubrible | ¿Se encuentra fácilmente? | Aparece en catálogo con descripción |
| Confiable | ¿Calidad monitoreada? | % tests de calidad verdes |
| Interoperable | ¿Formato semántico claro? | Uso de glosario y taxonomía |
| Accesible | ¿Permisos y SLA definidos? | Tiempo de provisión de acceso |
| Observado | ¿Métricas operacionales? | Freshness < 15 min |
---
## 3. Data Fabric (Concepto vs Mesh)
Data Fabric enfatiza las capacidades tecnológicas unificadas: catálogo, metadata activa, lineage, integración inteligente y automatización para descubrir, preparar y entregar datos sin fricción.

| Aspecto | Data Mesh | Data Fabric | Complemento |
|---------|-----------|-------------|-------------|
| Foco | Organización y modelo operativo | Plataforma y capacidades técnicas unificadas | Se combinan |
| Ownership | Dominios | Plataforma central + automatización | Mesh usa Fabric |
| Gobernanza | Federada | Policies/metadata activas | Fabric implementa políticas |
| Semántica | Glosarios/contratos | Catálogo semántico/grafo | Fabric potencia semantic layer |
---
## 4. Capa Semántica y Knowledge Graph (Fundamentos)
La capa semántica conecta datos dispersos mediante definiciones consistentes (conceptos, KPIs, dimensiones). Un **Knowledge Graph** modela entidades y relaciones, soportando descubrimiento y razonamiento.

Beneficios: búsqueda semántica, alineación terminológica, generación de consultas unificada, impact analysis y reutilización.

### Elementos Clave
- Conceptos (Customer, Order, Channel)
- Relaciones (Customer -> Order)
- Atributos normalizados (order_amount)
- KPIs derivados (LTV, Churn Rate)
- Mapeo a tablas/columnas físicas

---
## 5. Contratos de Datos (Mini Visión)
Los contratos definen: esquema, tipos, semántica, SLAs (latencia/frescura), calidad mínima, políticas de acceso. Sirven como interfaz estable entre productores y consumidores.

## 9. Resumen
Data Mesh aborda organización y responsabilidad; Data Fabric provee las capacidades técnicas de tejido inteligente; la capa semántica une el lenguaje y la interoperabilidad. Juntos permiten escalar valor sin perder control.

````python
# Ejemplo simple: representar metadatos de 2 productos de datos y validarlos
from dataclasses import dataclass, field
from typing import List, Dict

@dataclass
class DataField:
    name: str
    dtype: str
    description: str

@dataclass
class DataProduct:
    domain: str
    name: str
    owner: str
    fields: List[DataField]
    quality_rules: Dict[str, float]  # regla -> umbral mínimo
    sla_freshness_minutes: int

    def validate(self):
        problems = []
        if not self.owner:
            problems.append('Missing owner')
        if self.sla_freshness_minutes > 60:
            problems.append('Freshness SLA demasiado laxo (>60m)')
        for f in self.fields:
            if not f.description:
                problems.append(f'Campo {f.name} sin descripción')
        return problems

# Definimos dos productos de datos
ventas = DataProduct(
    domain='sales',
    name='orders_daily',
    owner='owner_sales',
    fields=[
        DataField('order_id','string','Identificador de pedido'),
        DataField('customer_id','string','Cliente único'),
        DataField('order_amount','decimal','Monto del pedido'),
        DataField('order_date','date','Fecha de la orden'),
        DataField('channel','string','Canal de venta')
    ],
    quality_rules={'completitud':0.99,'duplicados_max':0.001},
    sla_freshness_minutes=30
)

marketing = DataProduct(
    domain='marketing',
    name='campaign_performance',
    owner='owner_marketing',
    fields=[
        DataField('campaign_id','string','Identificador campaña'),
        DataField('channel','string','Canal de difusión'),
        DataField('impressions','int','Impresiones'),
        DataField('clicks','int','Clicks'),
        DataField('spend','decimal','Gasto invertido')
    ],
    quality_rules={'completitud':0.98,'duplicados_max':0.0005},
    sla_freshness_minutes=45
)

for dp in [ventas, marketing]:
    issues = dp.validate()
    print(f
 if not issues else f
````

## 3. Actualización 2025: Data Mesh, Data Fabric y Capa Semántica

### Definiciones Breves
- Data Mesh: Descentralización por dominios + data products con ownership y contratos.
- Data Fabric: Integración y orquestación habilitada por metadata y automatización sobre fuentes heterogéneas.
- Capa Semántica: Nivel lógico que define métricas y entidades uniformes (semantic layer / metric store).

### Estado 2025 (No Alucinaciones)
| Concepto | Madurez | Herramientas / Ejemplos |
|----------|---------|--------------------------|
| Data Mesh | En adopción selectiva | Data Contracts (JSON/YAML), Plataformas internas, DataHub para ownership |
| Data Fabric | Evolución en vendors | Integraciones con catálogos + orquestadores (Airflow/Dagster) |
| Semantic Layer | Creciendo | dbt Semantic Layer, Cube, Transform |

### Buenas Prácticas Iniciales (Nivel Junior)
1. Identificar 2–3 dominios reales (ej. Ventas, Clientes, Catálogo) en lugar de intentar cubrir toda la organización.
2. Definir para cada data product: propósito, esquema, contrato (campos, tipos, reglas, SLA disponibilidad).
3. Registrar ownership: equipo y responsable técnico (accountable).
4. Publicar métricas en capa semántica para evitar definiciones duplicadas.
5. Evitar crear complejidad (no introducir múltiples repos/infra si equipo pequeño).

### Ejemplo Simplificado de Contrato Data Product (YAML)
```
name: ventas_diarias
owner: equipo_ventas
slo:
  availability_hours: 2
  freshness_minutes: 60
schema:
  - field: fecha
    type: date
    quality_rule: not null
  - field: revenue
    type: decimal
    quality_rule: revenue >= 0
  - field: pedidos
    type: integer
    quality_rule: pedidos >= 0
lineage:
  sources:
    - sales_orders_raw
consumers:
  - dashboard_finanzas
```
