# Introducción a Cloud para Datos (Fundamentos)

## Objetivos
- Comprender diferencias cloud vs on-prem en el contexto de datos.
- Conocer servicios core: compute, storage, network, analytics.
- Identificar beneficios, riesgos iniciales y responsabilidades compartidas.
- Revisar modelos de servicio (IaaS, PaaS, SaaS) y de despliegue (single cloud, multi-cloud).

## 1. Cloud vs On-Prem para Datos
| Aspecto | On-Prem | Cloud |
|---------|--------|-------|
| CapEx/OpEx | Alta inversión inicial | Pago por uso |
| Escalado | Lento, compra hardware | Rápido, elástico |
| Time-to-market | Lento | Rápido |
| Seguridad | Control total | Modelo de responsabilidad compartida |
| Innovación | Limitada | Servicios gestionados (ML, streaming, etc.) |

## 2. Servicios Core
- Compute: EC2/VMs, Kubernetes, serverless (Lambda/Functions).
- Storage: objetos (S3/ADLS/GCS), bloques, archivos.
- Network: VPC/VNet, subnets, peering, NAT, egress.
- Analytics: BigQuery, Snowflake, Databricks, Redshift, Synapse.
- Mensajería/Streaming: Kafka MSK, Event Hubs, Pub/Sub, Kinesis.

## 3. Beneficios y Riesgos
**Beneficios:** elasticidad, servicios gestionados, innovación, disponibilidad global.
**Riesgos:** egress/costos variables, lock-in, seguridad/mal configuración, compliance jurisdiccional.

## 4. Responsabilidad Compartida
- Proveedor: seguridad de la nube.
- Cliente: seguridad en la nube (config, claves, datos, acceso).
- Implicaciones para datos: cifrado, IAM, tagging, auditoría, backup/DR.

## 5. Casos de Uso Típicos
- Data Warehouse gestionado (BigQuery/Snowflake).
- Lakehouse (Delta/Iceberg en object storage).
- Streaming/IoT (Kinesis/PubSub + Flink/Spark).
- ML gestionado (SageMaker/Vertex ML/AML).

## 8. Actualización 2025: Cloud para Datos

### Servicios Gestionados Clave
| Dominio | AWS | Azure | GCP |
|---------|-----|-------|-----|
| Ingestión | Kinesis / MSK | Event Hub | Pub/Sub |
| Procesamiento | EMR / Glue / Glue Studio | Azure Databricks / Synapse | Dataproc / Dataflow |
| Almacenamiento | S3 (Object) | ADLS Gen2 | GCS |
| Warehouse | Redshift | Synapse SQL / Azure SQL DW (legacy) | BigQuery |
| Lakehouse/Unified | Delta Lake (Databricks) | Delta en Databricks / Fabric Lakehouse | BigLake / BigQuery external + Iceberg |
| Catálogo / Governance | Glue Data Catalog / Lake Formation | Purview | Data Catalog |
| Streaming Analytics | Kinesis Analytics | Stream Analytics | Dataflow + BigQuery BI Engine |

### Tendencias 2025
- Lakehouse consolidado como patrón unificador (open formats + ACID + governance).
- Multi-cloud pragmático: distribución por regulación, costo y especialización (no replicar todo). 
- Aumento uso de formatos abiertos (Parquet + Iceberg/Delta/Hudi) para evitar lock-in.
- Observabilidad de costos (FinOps) integrada en dashboards internos.

### Buenas Prácticas Iniciales (Junior)
1. Separar storage (S3/ADLS/GCS) de compute para elasticidad.
2. Empezar con un proveedor principal; adoptar multi-cloud sólo por requerimiento concreto (regulatorio/performance).
3. Usar infraestructura como código mínima (Terraform) para reproducibilidad.
4. Activar encriptación por defecto y versionado de objetos críticos.
5. Elegir warehouse/lake según tipo de carga: analítica interactiva → BigQuery/Snowflake; mezcla ETL + ML → Databricks/Spark.

### Checklist de Inicio
- [ ] Cuenta/Subscription con política de naming.
- [ ] Roles IAM básicos (admin, data-engineer, read-only).
- [ ] Bucket/Container inicial con zonas (bronze/silver/gold).
- [ ] Catálogo base configurado.
- [ ] Pipeline ingestión batch simple (ej. CSV → Parquet).
