## 1. La Arquitectura de Tres Esquemas (ANSI/SPARC)

### Origen y Propósito

En 1975, el comité ANSI/SPARC (American National Standards Institute / Standards Planning and Requirements Committee) propuso una arquitectura de tres niveles para sistemas de bases de datos.

**Objetivo principal:** Lograr la **independencia de datos**, separando:
- Cómo el usuario ve los datos (vista externa)
- Cómo se estructura lógicamente (vista conceptual)
- Cómo se almacenan físicamente (vista interna)

### Los Tres Niveles

```
┌─────────────────────────────────────────┐
│     NIVEL EXTERNO (Vistas de Usuario)  │  ← ¿Qué ve cada usuario?
│   - Vista Ventas                        │
│   - Vista Inventario                    │
│   - Vista Finanzas                      │
└─────────────────┬───────────────────────┘
                  │ Mapeo Externo-Conceptual
┌─────────────────▼───────────────────────┐
│   NIVEL CONCEPTUAL (Modelo Lógico)     │  ← ¿Qué datos hay?
│   - Estructura global de la BD          │
│   - Entidades y Relaciones              │
│   - Restricciones de integridad         │
└─────────────────┬───────────────────────┘
                  │ Mapeo Conceptual-Interno
┌─────────────────▼───────────────────────┐
│   NIVEL INTERNO (Modelo Físico)        │  ← ¿Cómo se almacena?
│   - Estructuras de almacenamiento       │
│   - Índices, particiones                │
│   - Métodos de acceso                   │
└─────────────────────────────────────────┘
```

````python
# Visualización de la arquitectura de tres esquemas
import matplotlib.pyplot as plt
import matplotlib.patches as patches

fig, ax = plt.subplots(figsize=(12, 10))
ax.set_xlim(0, 10)
ax.set_ylim(0, 12)
ax.axis('off')

# Nivel Externo (usuarios)
usuarios = [
    {'x': 1, 'label': 'Analista\nVentas'},
    {'x': 3.5, 'label': 'Gerente\nInventario'},
    {'x': 6, 'label': 'Director\nFinanzas'},
    {'x': 8.5, 'label': 'Científico\nDatos'}
]

for user in usuarios:
    circle = plt.Circle((user['x'], 10.5), 0.4, color='#FF6B6B', alpha=0.7)
    ax.add_patch(circle)
    ax.text(user['x'], 10.5, '👤', fontsize=20, ha='center', va='center')
    ax.text(user['x'], 9.7, user['label'], fontsize=8, ha='center', va='top')

# Nivel Externo (Vistas)
rect_externo = patches.FancyBboxPatch((0.5, 7.5), 9, 1.5,
                                       boxstyle="round,pad=0.1",
                                       edgecolor='black', facecolor='#FFE66D',
                                       linewidth=2)
ax.add_patch(rect_externo)
ax.text(5, 8.7, 'NIVEL EXTERNO', fontsize=14, fontweight='bold', ha='center')
ax.text(5, 8.2, 'Vistas de Usuario (Esquemas Externos)', fontsize=10, ha='center', style='italic')
ax.text(5, 7.8, 'Vista_Ventas | Vista_Inventario | Vista_Finanzas | Vista_Analytics', 
        fontsize=8, ha='center')

# Flechas usuario -> externo
for user in usuarios:
    ax.annotate('', xy=(user['x'], 9), xytext=(user['x'], 9.7),
                arrowprops=dict(arrowstyle='->', lw=1.5, color='gray'))

# Mapeo Externo-Conceptual
ax.text(5, 7.1, '⬍ Mapeo Externo-Conceptual ⬍', fontsize=9, ha='center', 
        style='italic', color='#555')

# Nivel Conceptual
rect_conceptual = patches.FancyBboxPatch((0.5, 4.5), 9, 2,
                                          boxstyle="round,pad=0.1",
                                          edgecolor='black', facecolor='#4ECDC4',
                                          linewidth=2)
ax.add_patch(rect_conceptual)
ax.text(5, 6.2, 'NIVEL CONCEPTUAL', fontsize=14, fontweight='bold', ha='center')
ax.text(5, 5.7, 'Modelo Lógico Global (Esquema Conceptual)', fontsize=10, ha='center', style='italic')
ax.text(5, 5.2, 'Entidades: Cliente, Producto, Pedido, Categoría', fontsize=8, ha='center')
ax.text(5, 4.8, 'Relaciones: Cliente-Pedido, Pedido-Producto, Producto-Categoría', fontsize=8, ha='center')

# Flecha externo -> conceptual
ax.annotate('', xy=(5, 6.5), xytext=(5, 7.5),
            arrowprops=dict(arrowstyle='<->', lw=2, color='#555'))

# Mapeo Conceptual-Interno
ax.text(5, 4.1, '⬍ Mapeo Conceptual-Interno ⬍', fontsize=9, ha='center', 
        style='italic', color='#555')

# Nivel Interno (Físico)
rect_interno = patches.FancyBboxPatch((0.5, 1.5), 9, 2,
                                       boxstyle="round,pad=0.1",
                                       edgecolor='black', facecolor='#95E1D3',
                                       linewidth=2)
ax.add_patch(rect_interno)
ax.text(5, 3.2, 'NIVEL INTERNO', fontsize=14, fontweight='bold', ha='center')
ax.text(5, 2.7, 'Modelo Físico (Esquema Interno)', fontsize=10, ha='center', style='italic')
ax.text(5, 2.2, 'Almacenamiento: Tablas, Índices B-tree, Particiones por fecha', fontsize=8, ha='center')
ax.text(5, 1.8, 'Optimización: Clustering, Compresión, Estadísticas', fontsize=8, ha='center')

# Flecha conceptual -> interno
ax.annotate('', xy=(5, 3.5), xytext=(5, 4.5),
            arrowprops=dict(arrowstyle='<->', lw=2, color='#555'))

# Almacenamiento físico
ax.text(5, 0.8, '💾 Disco Físico', fontsize=12, ha='center', fontweight='bold')
ax.annotate('', xy=(5, 1.2), xytext=(5, 1.5),
            arrowprops=dict(arrowstyle='->', lw=2, color='#333'))

plt.title('Arquitectura de Tres Esquemas (ANSI/SPARC)', fontsize=16, fontweight='bold', pad=20)
plt.tight_layout()
plt.show()
````

## 2. Nivel Externo: Vistas de Usuario

### Concepto

El **nivel externo** representa cómo diferentes usuarios o grupos de usuarios ven los datos. Cada usuario tiene su propia "vista" personalizada.

### Características

- **Múltiples vistas:** Diferentes usuarios, diferentes perspectivas
- **Seguridad:** Cada usuario solo ve lo que necesita
- **Simplicidad:** Oculta la complejidad del modelo completo
- **Personalización:** Adaptado a necesidades específicas

### Ejemplos de Vistas Externas

````python
import pandas as pd

# Simulación de base de datos completa (nivel conceptual)
clientes_completo = pd.DataFrame({
    'cliente_id': [1, 2, 3],
    'nombre': ['Ana García', 'Luis Pérez', 'María López'],
    'email': ['ana@email.com', 'luis@email.com', 'maria@email.com'],
    'telefono': ['555-0001', '555-0002', '555-0003'],
    'direccion': ['Calle 1', 'Calle 2', 'Calle 3'],
    'credito_disponible': [5000, 3000, 7000],
    'historial_pagos': ['Bueno', 'Regular', 'Excelente']
})

print("="*80)
print("DATOS COMPLETOS (Nivel Conceptual)")
print("="*80)
print(clientes_completo)

# Vista 1: Analista de Ventas (solo necesita contacto)
vista_ventas = clientes_completo[['cliente_id', 'nombre', 'email', 'telefono']]

print("\n" + "="*80)
print("VISTA EXTERNA 1: Analista de Ventas")
print("="*80)
print(vista_ventas)

# Vista 2: Departamento de Crédito (necesita info financiera)
vista_credito = clientes_completo[['cliente_id', 'nombre', 'credito_disponible', 'historial_pagos']]

print("\n" + "="*80)
print("VISTA EXTERNA 2: Departamento de Crédito")
print("="*80)
print(vista_credito)

# Vista 3: Marketing (solo nombres y email, sin info sensible)
vista_marketing = clientes_completo[['nombre', 'email']]

print("\n" + "="*80)
print("VISTA EXTERNA 3: Marketing")
print("="*80)
print(vista_marketing)

print("\n💡 Cada departamento ve solo los datos que necesita para su trabajo")
````

## 3. Nivel Conceptual: Modelo Lógico Global

### Concepto

El **nivel conceptual** describe la estructura completa de la base de datos para toda la organización, independiente de cómo se almacena físicamente.

### Componentes del Modelo Conceptual

1. **Entidades:** Objetos del mundo real (Cliente, Producto, Pedido)
2. **Atributos:** Propiedades de las entidades
3. **Relaciones:** Asociaciones entre entidades
4. **Cardinalidades:** Cuántas instancias se relacionan
5. **Restricciones:** Reglas de negocio

### Ejemplo: Diagrama Entidad-Relación (ERD)

````python
# Representación visual de un modelo conceptual
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches

fig, ax = plt.subplots(figsize=(14, 8))
ax.set_xlim(0, 14)
ax.set_ylim(0, 8)
ax.axis('off')

# Definir entidades
entidades = [
    {'nombre': 'CLIENTE', 'x': 2, 'y': 6, 'atributos': ['cliente_id (PK)', 'nombre', 'email', 'teléfono']},
    {'nombre': 'PEDIDO', 'x': 7, 'y': 6, 'atributos': ['pedido_id (PK)', 'fecha', 'total', 'estado']},
    {'nombre': 'PRODUCTO', 'x': 12, 'y': 6, 'atributos': ['producto_id (PK)', 'nombre', 'precio', 'stock']},
    {'nombre': 'CATEGORÍA', 'x': 12, 'y': 2, 'atributos': ['categoria_id (PK)', 'nombre', 'descripción']}
]

# Dibujar entidades
for ent in entidades:
    # Rectángulo de la entidad
    rect = mpatches.Rectangle((ent['x']-1, ent['y']-0.3), 2, 0.6,
                               facecolor='#FFD93D', edgecolor='black', linewidth=2)
    ax.add_patch(rect)
    ax.text(ent['x'], ent['y'], ent['nombre'], fontsize=11, fontweight='bold',
            ha='center', va='center')
    
    # Atributos
    for i, attr in enumerate(ent['atributos']):
        ax.text(ent['x'], ent['y']-0.6-(i*0.25), attr, fontsize=8,
                ha='center', va='top', style='italic')

# Relaciones
# Cliente -> Pedido (1:N)
ax.plot([3, 6], [6, 6], 'k-', linewidth=2)
ax.text(4.5, 6.3, 'realiza', fontsize=10, ha='center', style='italic')
ax.text(3.3, 6.3, '1', fontsize=10, ha='center', fontweight='bold')
ax.text(5.7, 6.3, 'N', fontsize=10, ha='center', fontweight='bold')

# Pedido -> Producto (N:M a través de DetallePedido)
ax.plot([8, 11], [6, 6], 'k-', linewidth=2)
ax.text(9.5, 6.3, 'contiene', fontsize=10, ha='center', style='italic')
ax.text(8.3, 6.3, 'N', fontsize=10, ha='center', fontweight='bold')
ax.text(10.7, 6.3, 'M', fontsize=10, ha='center', fontweight='bold')

# Producto -> Categoría (N:1)
ax.plot([12, 12], [5.4, 2.6], 'k-', linewidth=2)
ax.text(12.5, 4, 'pertenece a', fontsize=10, ha='left', style='italic')
ax.text(12.3, 5.2, 'N', fontsize=10, ha='center', fontweight='bold')
ax.text(12.3, 2.8, '1', fontsize=10, ha='center', fontweight='bold')

# Leyenda
ax.text(7, 0.5, 'Cardinalidades: 1 (uno) | N (muchos) | M (muchos)', 
        fontsize=9, ha='center', bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))

plt.title('Modelo Conceptual (ERD) - Sistema de Ventas', fontsize=14, fontweight='bold', pad=20)
plt.tight_layout()
plt.show()

print("\n📌 Interpretación del Diagrama:")
print("   • Un CLIENTE puede realizar MUCHOS pedidos (1:N)")
print("   • Un PEDIDO puede contener MUCHOS productos y un PRODUCTO puede estar en MUCHOS pedidos (N:M)")
print("   • Un PRODUCTO pertenece a UNA categoría, una CATEGORÍA tiene MUCHOS productos (N:1)")
````

### Modelo Lógico Detallado

El modelo lógico es una refinación del conceptual, donde especificamos:
- Tipos de datos específicos
- Llaves primarias (PK) y foráneas (FK)
- Restricciones de integridad
- Normalización aplicada

````python
# Especificación del modelo lógico
modelo_logico = {
    'CLIENTE': {
        'atributos': [
            {'nombre': 'cliente_id', 'tipo': 'INT', 'pk': True, 'null': False},
            {'nombre': 'nombre', 'tipo': 'VARCHAR(100)', 'pk': False, 'null': False},
            {'nombre': 'email', 'tipo': 'VARCHAR(100)', 'pk': False, 'null': False, 'unique': True},
            {'nombre': 'telefono', 'tipo': 'VARCHAR(20)', 'pk': False, 'null': True},
            {'nombre': 'fecha_registro', 'tipo': 'TIMESTAMP', 'pk': False, 'null': False}
        ]
    },
    'PEDIDO': {
        'atributos': [
            {'nombre': 'pedido_id', 'tipo': 'INT', 'pk': True, 'null': False},
            {'nombre': 'cliente_id', 'tipo': 'INT', 'pk': False, 'null': False, 'fk': 'CLIENTE(cliente_id)'},
            {'nombre': 'fecha_pedido', 'tipo': 'TIMESTAMP', 'pk': False, 'null': False},
            {'nombre': 'total', 'tipo': 'DECIMAL(10,2)', 'pk': False, 'null': False},
            {'nombre': 'estado', 'tipo': 'VARCHAR(20)', 'pk': False, 'null': False}
        ]
    },
    'DETALLE_PEDIDO': {
        'atributos': [
            {'nombre': 'detalle_id', 'tipo': 'INT', 'pk': True, 'null': False},
            {'nombre': 'pedido_id', 'tipo': 'INT', 'pk': False, 'null': False, 'fk': 'PEDIDO(pedido_id)'},
            {'nombre': 'producto_id', 'tipo': 'INT', 'pk': False, 'null': False, 'fk': 'PRODUCTO(producto_id)'},
            {'nombre': 'cantidad', 'tipo': 'INT', 'pk': False, 'null': False},
            {'nombre': 'precio_unitario', 'tipo': 'DECIMAL(10,2)', 'pk': False, 'null': False}
        ]
    }
}

print("="*100)
print("MODELO LÓGICO - ESPECIFICACIÓN DETALLADA")
print("="*100)

for tabla, detalles in modelo_logico.items():
    print(f"\n📋 Tabla: {tabla}")
    print("-" * 100)
    print(f"{'Atributo':<20} {'Tipo':<20} {'PK':<5} {'NULL':<6} {'Otras Restricciones'}")
    print("-" * 100)
    
    for attr in detalles['atributos']:
        pk = '✓' if attr.get('pk', False) else ''
        null = 'NO' if not attr.get('null', True) else 'YES'
        otras = []
        if attr.get('unique'):
            otras.append('UNIQUE')
        if attr.get('fk'):
            otras.append(f"FK → {attr['fk']}")
        
        print(f"{attr['nombre']:<20} {attr['tipo']:<20} {pk:<5} {null:<6} {', '.join(otras)}")
````

## 4. Normalización: Estructurando Datos Coherentes

### ¿Qué es la Normalización?

La **normalización** es el proceso de organizar datos para:
- ✅ Eliminar redundancia
- ✅ Evitar anomalías de inserción, actualización y eliminación
- ✅ Garantizar integridad de datos
- ✅ Alinear estructura con lógica de negocio

### Las Tres Formas Normales Principales

#### Primera Forma Normal (1FN)
**Regla:** Cada atributo debe contener solo valores atómicos (indivisibles), sin grupos repetitivos.

#### Segunda Forma Normal (2FN)
**Regla:** Debe estar en 1FN + todos los atributos no-clave deben depender de la clave primaria completa.

#### Tercera Forma Normal (3FN)
**Regla:** Debe estar en 2FN + no debe haber dependencias transitivas (atributos que dependen de otros atributos no-clave).

````python
# Ejemplo: Proceso de Normalización
import pandas as pd

# Tabla NO normalizada (problemas múltiples)
print("="*100)
print("❌ TABLA SIN NORMALIZAR (Problemas de diseño)")
print("="*100)

sin_normalizar = pd.DataFrame({
    'pedido_id': [1, 2, 3],
    'cliente_nombre': ['Ana García', 'Luis Pérez', 'Ana García'],
    'cliente_email': ['ana@email.com', 'luis@email.com', 'ana@email.com'],
    'cliente_telefono': ['555-0001', '555-0002', '555-0001'],
    'productos': ['Laptop, Mouse', 'Teclado', 'Monitor, Webcam, Mouse'],  # ❌ Valores múltiples
    'precios': ['899.99, 29.99', '79.99', '299.99, 59.99, 29.99'],  # ❌ Valores múltiples
    'categoria_producto': ['Electrónica', 'Electrónica', 'Electrónica'],  # ❌ Redundancia
})

print(sin_normalizar)
print("\n⚠️ Problemas identificados:")
print("   1. Información de cliente repetida (redundancia)")
print("   2. Valores múltiples en 'productos' y 'precios' (viola 1FN)")
print("   3. Si Ana cambia su email, hay que actualizar múltiples registros (anomalía de actualización)")
print("   4. No se pueden agregar productos sin pedido (anomalía de inserción)")
````

````python
# Aplicando 1FN: Eliminar grupos repetitivos
print("\n" + "="*100)
print("✅ PRIMERA FORMA NORMAL (1FN): Valores atómicos")
print("="*100)

primera_fn = pd.DataFrame({
    'pedido_id': [1, 1, 2, 3, 3, 3],
    'cliente_nombre': ['Ana García', 'Ana García', 'Luis Pérez', 'Ana García', 'Ana García', 'Ana García'],
    'cliente_email': ['ana@email.com', 'ana@email.com', 'luis@email.com', 'ana@email.com', 'ana@email.com', 'ana@email.com'],
    'cliente_telefono': ['555-0001', '555-0001', '555-0002', '555-0001', '555-0001', '555-0001'],
    'producto': ['Laptop', 'Mouse', 'Teclado', 'Monitor', 'Webcam', 'Mouse'],
    'precio': [899.99, 29.99, 79.99, 299.99, 59.99, 29.99],
    'categoria': ['Electrónica', 'Electrónica', 'Electrónica', 'Electrónica', 'Electrónica', 'Electrónica']
})

print(primera_fn)
print("\n✅ Logrado: Todos los valores son atómicos")
print("⚠️ Todavía hay redundancia: Datos de cliente repetidos")
````

````python
# Aplicando 2FN: Eliminar dependencias parciales
print("\n" + "="*100)
print("✅ SEGUNDA FORMA NORMAL (2FN): Separar entidades")
print("="*100)

# Tabla Clientes (datos de cliente separados)
clientes_2fn = pd.DataFrame({
    'cliente_id': [1, 2],
    'nombre': ['Ana García', 'Luis Pérez'],
    'email': ['ana@email.com', 'luis@email.com'],
    'telefono': ['555-0001', '555-0002']
})

# Tabla Pedidos
pedidos_2fn = pd.DataFrame({
    'pedido_id': [1, 2, 3],
    'cliente_id': [1, 2, 1],
    'fecha': ['2024-11-01', '2024-11-02', '2024-11-03']
})

# Tabla Detalle Pedidos
detalle_2fn = pd.DataFrame({
    'detalle_id': [1, 2, 3, 4, 5, 6],
    'pedido_id': [1, 1, 2, 3, 3, 3],
    'producto': ['Laptop', 'Mouse', 'Teclado', 'Monitor', 'Webcam', 'Mouse'],
    'precio': [899.99, 29.99, 79.99, 299.99, 59.99, 29.99],
    'categoria': ['Electrónica', 'Electrónica', 'Electrónica', 'Electrónica', 'Electrónica', 'Electrónica']
})

print("\n📋 Tabla CLIENTES:")
print(clientes_2fn)
print("\n📋 Tabla PEDIDOS:")
print(pedidos_2fn)
print("\n📋 Tabla DETALLE_PEDIDOS:")
print(detalle_2fn)

print("\n✅ Logrado: Datos de cliente ya no se repiten")
print("⚠️ Todavía hay redundancia: Nombre de producto y categoría repetidos")
````

````python
# Aplicando 3FN: Eliminar dependencias transitivas
print("\n" + "="*100)
print("✅ TERCERA FORMA NORMAL (3FN): Modelo completamente normalizado")
print("="*100)

# Tabla Clientes (sin cambios)
clientes_3fn = clientes_2fn.copy()

# Tabla Pedidos (sin cambios)
pedidos_3fn = pedidos_2fn.copy()

# Tabla Productos (nueva)
productos_3fn = pd.DataFrame({
    'producto_id': [1, 2, 3, 4, 5],
    'nombre': ['Laptop', 'Mouse', 'Teclado', 'Monitor', 'Webcam'],
    'categoria_id': [1, 1, 1, 1, 1]
})

# Tabla Categorías (nueva)
categorias_3fn = pd.DataFrame({
    'categoria_id': [1],
    'nombre': ['Electrónica'],
    'descripcion': ['Dispositivos electrónicos y accesorios']
})

# Tabla Detalle Pedidos (actualizada con referencias)
detalle_3fn = pd.DataFrame({
    'detalle_id': [1, 2, 3, 4, 5, 6],
    'pedido_id': [1, 1, 2, 3, 3, 3],
    'producto_id': [1, 2, 3, 4, 5, 2],
    'cantidad': [1, 1, 1, 1, 1, 1],
    'precio_unitario': [899.99, 29.99, 79.99, 299.99, 59.99, 29.99]
})

print("\n📋 Tabla CLIENTES:")
print(clientes_3fn)
print("\n📋 Tabla PEDIDOS:")
print(pedidos_3fn)
print("\n📋 Tabla PRODUCTOS:")
print(productos_3fn)
print("\n📋 Tabla CATEGORÍAS:")
print(categorias_3fn)
print("\n📋 Tabla DETALLE_PEDIDOS:")
print(detalle_3fn)

print("\n✅ ¡Modelo completamente normalizado (3FN)!")
print("\n📊 Beneficios logrados:")
print("   • Cero redundancia de datos")
print("   • Integridad referencial garantizada")
print("   • Fácil mantenimiento")
print("   • Estructura alineada con lógica de negocio")
print("   • Sin anomalías de actualización, inserción o eliminación")
````

## 5. Nivel Interno: Modelo Físico

### Concepto

El **nivel interno** describe cómo los datos se almacenan físicamente en el disco, incluyendo:
- Estructuras de almacenamiento
- Índices para optimizar búsquedas
- Particiones de datos
- Estrategias de compresión
- Métodos de acceso

### Del Modelo Lógico al Físico

````python
# Ejemplo: Implementación física (SQL DDL)
implementacion_fisica = """
-- ============================================================
-- MODELO FÍSICO: Implementación en PostgreSQL
-- ============================================================

-- Tabla CLIENTES
CREATE TABLE clientes (
    cliente_id      SERIAL PRIMARY KEY,
    nombre          VARCHAR(100) NOT NULL,
    email           VARCHAR(100) NOT NULL UNIQUE,
    telefono        VARCHAR(20),
    fecha_registro  TIMESTAMP DEFAULT CURRENT_TIMESTAMP,
    
    -- Índices para optimizar búsquedas
    INDEX idx_cliente_email (email),
    INDEX idx_cliente_nombre (nombre)
) 
TABLESPACE tbs_clientes;  -- Almacenamiento físico específico

-- Tabla PEDIDOS (particionada por fecha)
CREATE TABLE pedidos (
    pedido_id       SERIAL,
    cliente_id      INTEGER NOT NULL,
    fecha_pedido    TIMESTAMP NOT NULL,
    total           DECIMAL(10,2) NOT NULL,
    estado          VARCHAR(20) NOT NULL,
    
    PRIMARY KEY (pedido_id, fecha_pedido),
    FOREIGN KEY (cliente_id) REFERENCES clientes(cliente_id)
        ON DELETE RESTRICT
        ON UPDATE CASCADE,
    
    -- Índices
    INDEX idx_pedido_cliente (cliente_id),
    INDEX idx_pedido_fecha (fecha_pedido),
    INDEX idx_pedido_estado (estado)
)
PARTITION BY RANGE (fecha_pedido);  -- Particionamiento por rango de fechas

-- Crear particiones por mes
CREATE TABLE pedidos_2024_11 PARTITION OF pedidos
    FOR VALUES FROM ('2024-11-01') TO ('2024-12-01')
    TABLESPACE tbs_pedidos_2024;

CREATE TABLE pedidos_2024_12 PARTITION OF pedidos
    FOR VALUES FROM ('2024-12-01') TO ('2025-01-01')
    TABLESPACE tbs_pedidos_2024;

-- Tabla PRODUCTOS con compresión
CREATE TABLE productos (
    producto_id     SERIAL PRIMARY KEY,
    nombre          VARCHAR(200) NOT NULL,
    descripcion     TEXT,
    precio          DECIMAL(10,2) NOT NULL,
    stock           INTEGER NOT NULL DEFAULT 0,
    categoria_id    INTEGER NOT NULL,
    
    FOREIGN KEY (categoria_id) REFERENCES categorias(categoria_id),
    
    -- Índice B-tree para búsqueda por nombre
    INDEX idx_producto_nombre USING BTREE (nombre),
    
    -- Índice para búsquedas por rango de precio
    INDEX idx_producto_precio (precio),
    
    -- Índice compuesto para consultas frecuentes
    INDEX idx_producto_cat_precio (categoria_id, precio)
)
WITH (fillfactor = 80);  -- Optimización de almacenamiento

-- Estadísticas para el optimizador
ANALYZE clientes;
ANALYZE pedidos;
ANALYZE productos;
"""

print("="*100)
print("MODELO FÍSICO: Implementación SQL con Optimizaciones")
print("="*100)
print(implementacion_fisica)

print("\n💾 Aspectos del Modelo Físico:")
print("   • Tipos de datos específicos del DBMS (SERIAL, VARCHAR, DECIMAL)")
print("   • Índices B-tree para búsquedas eficientes")
print("   • Particionamiento por fecha (mejor rendimiento en consultas temporales)")
print("   • Tablespaces para control de almacenamiento físico")
print("   • Factores de llenado para optimizar inserciones")
print("   • Estadísticas para el optimizador de consultas")
````

## 6. Beneficios de la Arquitectura de Tres Esquemas

### 1. Independencia de Datos

**Independencia Lógica:**
- Cambios en el esquema conceptual no afectan a las vistas externas
- Ejemplo: Agregar una nueva tabla no rompe aplicaciones existentes

**Independencia Física:**
- Cambios en el almacenamiento no afectan el esquema conceptual
- Ejemplo: Agregar índices o cambiar particiones es transparente

### 2. Seguridad
- Cada usuario ve solo sus datos autorizados
- Nivel externo actúa como capa de seguridad

### 3. Flexibilidad
- Múltiples vistas para diferentes necesidades
- Fácil adaptación a cambios de negocio

### 4. Mantenibilidad
- Cambios localizados en cada nivel
- Menor riesgo de efectos secundarios

### 5. Optimización
- Nivel físico puede optimizarse sin afectar la lógica
- Mejor rendimiento sin cambiar aplicaciones

## 8. Resumen y Puntos Clave

### Arquitectura de Tres Esquemas

1. **Nivel Externo (Vistas)**
   - Cómo cada usuario ve los datos
   - Múltiples vistas personalizadas
   - Seguridad y simplicidad

2. **Nivel Conceptual (Lógico)**
   - Estructura global de la base de datos
   - Independiente de la implementación

   - Índices, particiones, compresión
   - Optimizaciones de rendimiento

### Normalización

- **1FN:** Valores atómicos, sin grupos repetitivos
- **2FN:** 1FN + dependencia total de la clave primaria
- **3FN:** 2FN + sin dependencias transitivas

✅ Independencia de datos (lógica y física)

✅ Eliminación de redundancia

## 22. Actualización 2025: Star vs Snowflake vs Data Vault

### Panorama
La selección del esquema depende de requerimientos de auditabilidad, flexibilidad de cambios en negocio, rendimiento de BI y volumen/variedad de datos. En 2025 conviven modelos dimensionales optimizados con enfoques Data Vault para historización granular y linaje auditable, mientras algunos equipos añaden capas semánticas (metric stores) encima para definición única de métricas.

### Comparativa
| Criterio | Star Schema | Snowflake | Data Vault |
|----------|-------------|-----------|------------|
| Simplicidad Consulta | Alta | Media | Baja |
| Redundancia Dimensiones | Alta (desnormalización controlada) | Menor | Baja (estructura hub/link/satellite) |
| Rendimiento BI | Muy bueno (menos joins) | Bueno | Variable (más joins) |
| Mantenimiento Cambios | Medio (añadir atributos) | Medio-Alto | Bueno (satélites separados) |
| Auditoría / Historización | Limitada (SCD manual) | Mejor con SCD | Nativa (satélites históricos) |
| Time-to-Market | Rápido | Medio | Más lento inicial |
| Complejidad Modelado | Baja | Media | Alta |
| Evolución Esquema | Media | Media-Alta | Alta (satélites nuevos) |
| Uso Típico | Dashboards, métricas conformadas | BI corporativo con muchas dimensiones | Regulación, trazabilidad, múltiples fuentes heterogéneas |

### Buenas Prácticas 2025
- Star + Data Vault híbrido: diseñar Vault para ingestión/auditoría y publicar Data Marts dimensionales derivados.
- Evitar snowflake excesivo que fragmenta lectura (solo normalizar subdimensiones de baja cardinalidad o que cambian frecuentemente).
- Usar SCD Tipo 2 en dimensiones críticas (cliente, producto) cuando impacto analítico histórico es importante.
- Catálogo y linaje automático para rastrear derivaciones (DataHub/OpenMetadata + OpenLineage).
