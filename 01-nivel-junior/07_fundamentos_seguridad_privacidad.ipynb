## 1. ¿Por qué Seguridad de Datos?

**Seguridad de datos:** Proteger información contra acceso no autorizado, corrupción o robo.

**Riesgos comunes:**
- Acceso no autorizado (interno o externo).
- Filtración de datos sensibles (breaches).
- Modificación maliciosa o accidental.
- Pérdida de datos (sin backups).

**Impacto:**
- Legal: Multas (GDPR hasta 4% ingresos anuales).
- Reputacional: Pérdida de confianza.
- Operacional: Interrupción de servicios.

## 2. Tipos de Datos Sensibles

| Tipo | Ejemplos | Riesgo |
|------|----------|--------|
| **PII (Personal Identifiable Information)** | Nombre, email, DNI, dirección, teléfono | Identifica a personas, sujeto a GDPR/CCPA |
| **PCI (Payment Card Industry)** | Número tarjeta, CVV | Fraude financiero |
| **PHI (Protected Health Information)** | Historial médico, diagnósticos | Privacidad salud (HIPAA) |
| **Propiedad Intelectual** | Código fuente, algoritmos, estrategias | Ventaja competitiva |
| **Datos de Menores** | Información de <13 años (COPPA) | Protección reforzada |

**Clasificación:** Público, Interno, Confidencial, Regulado (visto en módulos previos).

## 3. Principios de Privacidad

Basados en **GDPR** (General Data Protection Regulation) y normativas similares:

1. **Legalidad y transparencia:** Informar al titular sobre el uso de sus datos.
2. **Limitación de finalidad:** Usar datos solo para el propósito declarado.
3. **Minimización:** Recopilar solo datos necesarios.
4. **Exactitud:** Mantener datos actualizados y precisos.
5. **Limitación de almacenamiento:** Retener solo el tiempo necesario.
6. **Integridad y confidencialidad:** Proteger contra accesos no autorizados.
7. **Responsabilidad proactiva:** Demostrar cumplimiento (accountability).

## 4. Derechos de los Titulares (GDPR y equivalentes)

| Derecho | Descripción |
|---------|-------------|
| **Acceso** | Saber qué datos se tienen y cómo se usan |
| **Rectificación** | Corregir datos inexactos |
| **Supresión (olvido)** | Borrar datos bajo ciertas condiciones |
| **Portabilidad** | Recibir datos en formato estructurado y transferible |
| **Oposición** | Oponerse al procesamiento (ej. marketing) |
| **Restricción** | Limitar el procesamiento temporalmente |

**Implicación técnica:** Sistemas deben permitir localizar, exportar y borrar datos de un usuario específico.

## 5. Controles Básicos de Seguridad

```python
import pandas as pd
import hashlib

# Ejemplo: dataset con PII
df = pd.DataFrame({
    'user_id': [1, 2, 3],
    'nombre': ['Ana García', 'Luis Pérez', 'María López'],
    'email': ['ana@example.com', 'luis@example.com', 'maria@example.com'],
    'saldo': [1500, 2300, 800]
});

print('Dataset original (con PII):')
print(df)

# Control 1: Pseudonimización (hashear email)
df_pseudo = df.copy()
df_pseudo['email_hash'] = df_pseudo['email'].apply(lambda x: hashlib.sha256(x.encode()).hexdigest()[:16])
df_pseudo = df_pseudo.drop(columns=['email'])

print('\nDataset pseudonimizado:')
print(df_pseudo)

# Control 2: Anonimización (eliminar identificadores)
df_anon = df[['saldo']].copy()
print('\nDataset anonimizado (solo agregados):')
print(df_anon.describe())
```

## 6. Conceptos Clave de Seguridad

### 6.1. Autenticación vs Autorización
- **Autenticación:** Verificar identidad (¿quién eres?). Ej: usuario/contraseña, SSO, MFA.
- **Autorización:** Verificar permisos (¿qué puedes hacer?). Ej: RBAC (roles), ABAC (atributos).

### 6.2. Encriptación (Cifrado)
- **En reposo (at-rest):** Datos almacenados cifrados (disco, S3, DB). Ej: AES-256.
- **En tránsito (in-transit):** Datos cifrados al transmitirse. Ej: TLS/HTTPS.
- **Clave de cifrado:** Gestionada por KMS (Key Management Service).

### 6.3. Control de Acceso
- **Principio de mínimo privilegio:** Solo dar acceso necesario.
- **Segregación de funciones:** Separar responsabilidades críticas.
- **Revisión periódica:** Auditar permisos (ej. cada 90 días).

## 7. Amenazas Comunes

| Amenaza | Descripción | Mitigación |
|---------|-------------|------------|
| **Phishing** | Engaño para robar credenciales | Capacitación, MFA |
| **Inyección SQL** | Manipular consultas | Parametrización, validación |
| **Acceso interno malicioso** | Usuario con acceso legítimo roba datos | RBAC, auditoría, DLP |
| **Ransomware** | Cifrado malicioso + extorsión | Backups aislados, segmentación |
| **Exposición de credenciales** | Contraseñas en código | Secrets management, rotación |
