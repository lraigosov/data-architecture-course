# Fundamentos de Arquitectura de Datos en Tiempo Real, Edge & Streaming

**Objetivo:** Comprender diferencias entre micro-batch y streaming continuo, principios básicos de plataformas de streaming y motivación del modelo edge-cloud.

---
## 1. ¿Por qué Tiempo Real?
Casos: fraude, mantenimiento predictivo, recomendaciones dinámicas, IoT sensores, monitoreo operaciones. La latencia aceptable puede ir de milisegundos a segundos.

| Caso | Latencia Objetivo | Ejemplo Métrica |
|------|------------------|-----------------|
| Detección de fraude | < 200 ms | Tiempo decisión |
| Mantenimiento predictivo | < 5 s | Tiempo alerta |
| Recomendación e-commerce | < 300 ms | Tiempo respuesta API |
| Telemetría IoT | 1–10 s | Lag ingestión |

---
## 2. Micro-batch vs Streaming Continuo
| Aspecto | Micro-batch | Streaming Continuo |
|---------|-------------|-------------------|
| Latencia | Segundos-minutos | Milisegundos-segundos |
| Complejidad | Menor | Mayor (estado, orden, watermarks) |
| Semántica | Tandas fijas | Evento a evento |
| Casos de uso | Aggregaciones periódicas | Alertas, scoring online |
| Herramientas | Spark Structured Streaming (trigger), dbt incremental | Flink, Kafka Streams, Pulsar Functions |

---
## 3. Plataformas de Streaming (Visión)
Kafka: log distribuido, particiones, offsets, consumo pub/sub.
Pulsar: separación storage/servicio, multi-tenancy, funciones integradas.
Kinesis/EventHub/PubSub: servicios administrados cloud.

Conceptos clave: particiones, offset, consumer group, retención, semántica de entrega (at-least-once, exactly-once), watermark, ventana (tumbling, sliding), estado.

---
## 4. Edge vs Cloud (Modelo Distribuido)
Edge: cercanía a la fuente (gateway, dispositivo) para reducir latencia, pre-filtrar, comprimir, anonimizar.
Cloud: cómputo escalable, almacenamiento duradero, modelos ML pesados, agregaciones históricas.

Beneficios edge: menor latencia, menor ancho de banda, resiliencia offline parcial.
Riesgos: complejidad operativa, actualización de software, seguridad física.
