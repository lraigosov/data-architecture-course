## 1. ¿Qué es una Cultura Data-Driven?
Una organización **data-driven** toma decisiones basadas en datos y análisis, no solo en intuición o experiencia.

### Características
- **Acceso democratizado:** Todos los empleados pueden acceder a datos relevantes para su rol.
- **Confianza en datos:** Los datos son precisos, actualizados y bien documentados.
- **Decisiones respaldadas:** Cada decisión importante se fundamenta con evidencia cuantitativa.
- **Experimentación:** Se valida con pruebas (A/B testing, pilotos) antes de escalar.
- **Aprendizaje continuo:** Se revisan métricas para iterar y mejorar.

### Ejemplo: Retail Tradicional vs Data-Driven
| Decisión                | Tradicional                          | Data-Driven                                    |
|------------------------|--------------------------------------|-----------------------------------------------|
| Inventario de producto | "Este producto siempre se vende bien" | Análisis de rotación + estacionalidad + forecast |
| Precio de oferta       | "Bajemos 20% como siempre"           | Elasticidad de precio + margen objetivo       |
| Apertura de tienda     | "En esa zona hay tráfico"            | Análisis geoespacial + perfil demográfico     |
| Contratación           | "Necesitamos más vendedores"         | Correlación ventas/empleado + picos de demanda |

## 2. Beneficios de una Cultura Data-Driven

### Para la Organización
1. **Mejor toma de decisiones:** Reducción de errores costosos (ej: lanzar producto sin demanda).
2. **Eficiencia operativa:** Identificar cuellos de botella, optimizar procesos.
3. **Innovación acelerada:** Probar hipótesis rápidamente con experimentos.
4. **Ventaja competitiva:** Responder más rápido a cambios del mercado.
5. **Transparencia:** Métricas compartidas alinean equipos hacia objetivos comunes.

### Para Empleados
- **Autonomía:** Pueden tomar decisiones sin esperar aprobación (si tienen datos).
- **Desarrollo profesional:** Adquieren habilidades de análisis valiosas.
- **Reconocimiento:** Contribuciones basadas en impacto medible.

### Caso Real: Netflix
- **Decisión:** Producir "House of Cards" (2013).
- **Datos usados:** 
  - Usuarios que vieron películas del director David Fincher.
  - Fans de Kevin Spacey.
  - Popularidad de la serie original británica.
- **Resultado:** Serie exitosa, validó modelo de producción basado en datos.

## 3. Barreras Comunes para Adopción

### 1. Falta de Acceso
- **Problema:** Datos están en silos (cada departamento tiene su BD).
- **Solución:** Catálogo de datos centralizado, permisos claros.

### 2. Desconfianza en Datos
- **Problema:** "Estos números no pueden ser correctos".
- **Solución:** Validación de calidad, lineage (saber de dónde vienen los datos).

### 3. Falta de Habilidades (Data Literacy)
- **Problema:** Empleados no saben interpretar gráficos o estadísticas.
- **Solución:** Capacitación básica, dashboards intuitivos.

### 4. Resistencia al Cambio
- **Problema:** "Siempre lo hemos hecho así, no necesitamos datos".
- **Solución:** Quick wins (éxitos rápidos), sponsorship ejecutivo.

### 5. Datos de Baja Calidad
- **Problema:** Campos vacíos, duplicados, formatos inconsistentes.
- **Solución:** Procesos de limpieza, reglas de validación, data stewardship.

## 4. Data Literacy: Habilidades Fundamentales
**Data Literacy** = capacidad de leer, trabajar, analizar y comunicar con datos.

### Niveles de Data Literacy
| Nivel       | Habilidades                                      | Ejemplo                          |

| **Básico**  | Leer tablas, gráficos de barras/líneas          | Entender dashboard de ventas     |
| **Intermedio** | Filtrar datos, calcular promedios, detectar outliers | Analizar campaña de marketing |
| **Avanzado** | SQL, estadística básica, pruebas de hipótesis   | Diseñar experimento A/B          |
| **Experto**  | Modelado predictivo, machine learning           | Forecast de demanda              |

## 5. Ejemplos de Uso de Datos por Rol

### Rol: Vendedor
- **Pregunta:** ¿Qué clientes debo contactar hoy?
- **Datos:** Clientes con alta probabilidad de compra (scoring predictivo), sin contacto en 30 días.
- **Decisión:** Priorizar lista de 20 clientes top.

### Rol: Gerente de Marketing
- **Pregunta:** ¿Qué canal de adquisición tiene mejor ROI?
- **Datos:** Costo por lead (CPL) y tasa de conversión por canal (Google Ads, Facebook, email).
- **Decisión:** Reasignar 30% de presupuesto de canal con bajo ROI a canal ganador.

### Rol: Gerente de Operaciones
- **Pregunta:** ¿Dónde están los cuellos de botella en producción?
- **Datos:** Tiempo de ciclo por estación, tasa de defectos.
- **Decisión:** Invertir en automatización de estación con mayor variabilidad.

### Rol: CEO
- **Pregunta:** ¿Entramos al mercado de Brasil?
- **Datos:** Tamaño de mercado, competencia, costos logísticos, regulación.
- **Decisión:** Piloto en São Paulo por 6 meses con métrica: CAC < $50, LTV > $200.

## 6. Código: Simulador de Decisión Basada en Datos

````python
import random
from dataclasses import dataclass
from typing import List

@dataclass
class Producto:
    nombre: str
    rotacion_mensual: int  # Unidades vendidas/mes
    costo_inventario: float  # Costo de mantener 1 unidad en inventario
    margen_unitario: float
    
def calcular_inventario_optimo(producto: Producto, meses_forecast: int = 3) -> dict:
    """Decisión data-driven: cuánto inventario mantener."""
    demanda_esperada = producto.rotacion_mensual * meses_forecast
    costo_mantener = demanda_esperada * producto.costo_inventario * meses_forecast
    ingreso_potencial = demanda_esperada * producto.margen_unitario
    
    # Política: mantener inventario solo si margen > costo de inventario * 2
    if ingreso_potencial > costo_mantener * 2:
        decision = "Mantener inventario completo"
        unidades = demanda_esperada
    else:
        decision = "Inventario mínimo (JIT - Just In Time)"
        unidades = producto.rotacion_mensual  # Solo 1 mes
    
    return {
        "producto": producto.nombre,
        "decision": decision,
        "unidades_mantener": int(unidades),
        "costo_inventario": round(costo_mantener, 2),
        "ingreso_potencial": round(ingreso_potencial, 2),
        "roi": round(ingreso_potencial / costo_mantener if costo_mantener > 0 else 0, 2)
    }

# Ejemplo: Decisión de inventario para 3 productos
productos = [
    Producto("Laptop Gaming", 50, 5.0, 200),  # Alta rotación, alto margen
    Producto("Mouse básico", 200, 0.5, 3),    # Alta rotación, bajo margen
    Producto("Teclado mecánico", 10, 2.0, 30) # Baja rotación, margen medio
]

print("=== Decisiones de Inventario Data-Driven ===")
for p in productos:
    resultado = calcular_inventario_optimo(p)
    print(f"\n{resultado['producto']}:")
    print(f"  Decisión: {resultado['decision']}")
    print(f"  Unidades a mantener: {resultado['unidades_mantener']}")
    print(f"  ROI esperado: {resultado['roi']}x")

print("\n=== Comparación vs Intuición ===")
print("Decisión intuitiva: 'Mantener 3 meses de todos los productos'")
print("Decisión data-driven: 'Laptop completo, Mouse JIT (bajo margen), Teclado JIT (baja rotación)'")
print("Ahorro estimado: 40% en costos de inventario sin afectar disponibilidad.")
````

## 7. Primeros Pasos para Cultura Data-Driven

### Para Individuos
1. **Pregunta con datos:** Antes de opinar, busca evidencia.
2. **Aprende herramientas básicas:** Excel avanzado, Google Sheets, SQL básico.
3. **Cuestiona métricas:** ¿De dónde vienen? ¿Son actuales? ¿Completas?
4. **Comparte insights:** Presenta hallazgos con visualizaciones simples.

### Para Equipos
1. **Define 3-5 métricas clave (North Star):** Ej: NPS, Churn Rate, Revenue per User.
2. **Dashboard semanal:** Revisión de métricas en reuniones de equipo.
3. **Postmortems con datos:** Analizar qué funcionó/no funcionó en proyectos.
4. **Experimentos pequeños:** Probar cambios en muestra pequeña antes de escalar.

### Para Organizaciones (ver Senior)
- Patrocinio ejecutivo, inversión en plataforma de datos, programa de capacitación.
