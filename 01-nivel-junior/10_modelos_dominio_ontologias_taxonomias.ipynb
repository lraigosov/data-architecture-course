## 1. Definiciones Clave

| Concepto | Propósito | Estructura | Ejemplo |
|----------|-----------|-----------|---------|
| Dominio | Área de negocio con reglas propias | Entidades + reglas | Ventas, Marketing, Riesgos |
| Modelo de dominio | Representación conceptual de entidades y relaciones | Diagrama, clases | Cliente, Pedido, Factura |
| Taxonomía | Clasificación jerárquica de conceptos | Árbol (padre-hijo) | Producto → Electrónicos → Móviles |
| Ontología | Modelo semántico rico (conceptos + relaciones + restricciones) | Grafo (RDF/OWL) | Cliente OWL con relación "realizaPedido" |
| Glosario | Definiciones comunes y unívocas | Lista tabular | "MRR": Ingresos recurrentes mensuales |

### Diferencias rápidas
- **Taxonomía**: quién es padre de quién.
- **Ontología**: qué relaciones y semántica ("Cliente compra Producto").
- **Modelo de dominio**: entidades y reglas enfocadas en lógica.
- **Glosario**: vocabulario estándar para evitar ambigüedades.

## 2. ¿Por qué Importa Estratégicamente?

| Beneficio Estratégico | Cómo ayuda el modelo | Ejemplo |
|-----------------------|----------------------|---------|
| Alineación a KPIs | Define qué datos soportan cada indicador | KPI: Churn → entidades: Cliente, Cancelación |
| Gobernanza | Claridad de ownership por dominio | Dominio Facturación → Owner: Equipo Finanzas |
| Calidad | Reglas de negocio claras → validaciones precisas | TotalPedido = suma(líneas) |
| Reutilización | Evita duplicar definiciones en cada equipo | SegmentoCliente estándar |
| Integración | Facilita mapping entre fuentes heterogéneas | Campo "customer_id" unificado |
| Evolución | Cambios controlados con impacto visible | Nueva métrica MRR se agrega a Ontología |

### Señales de Problemas sin Modelos
- KPI "Ingresos" calculado de 3 maneras distintas.
- Columnas con prefijos ambiguos (cust_, cli_, customer_).
- Dificultad para saber quién es el dueño de un dataset crítico.
- Retrabajo constante en mapeos y transformaciones.

## 3. Introducción a Domain-Driven Design (DDD) aplicado a Datos

| Concepto DDD | Adaptación al mundo de datos |
|--------------|------------------------------|
| Bounded Context | Frontera donde una palabra tiene significado preciso ("Cliente" en Marketing vs Financiero) |
| Ubiquitous Language | Glosario compartido codificado en catálogos y contratos |
| Aggregates | Conjunto consistente de entidades (Pedido + Líneas) para integridad |
| Entities | Objetos con identidad estable (cliente_id) |
| Value Objects | Datos sin identidad propia (dirección, moneda) |
| Domain Events | Hechos relevantes (PedidoCreado, ClienteSegmentado) |

DDD reduce fricción semántica y soporta evolución incremental del ecosistema de datos.

## 4. Taxonomía Sencilla en Código
Representaremos una taxonomía de productos y la usaremos para clasificar registros.

````python
# Construcción de una taxonomía simple usando diccionarios anidados
taxonomia_productos = {
    'Electrónicos': {
        'Computadoras': ['Laptop', 'Desktop'],
        'Móviles': ['Teléfono', 'Tablet']
    },
    'Hogar': {
        'Cocina': ['Licuadora', 'Horno'],
        'Muebles': ['Silla', 'Mesa']
    }
}

def imprimir_taxonomia(taxonomia):
    for categoria, subcategorias in taxonomia.items():
        print(f'- {categoria}')
        for subcategoria, productos in subcategorias.items():
            print(f'  - {subcategoria}: {", ".join(productos)}')

imprimir_taxonomia(taxonomia_productos)
````
