
## 1. C4 Model: Niveles de Abstracción
El **C4 Model** (Context, Containers, Components, Code) estructura visualización en capas:

### Nivel 1: Contexto (System Context)
Vista más abstracta: sistema central + actores externos.
```mermaid
graph TD
    User[Usuario Final] --> DataPlatform[Plataforma de Datos]
    CRM[Sistema CRM] --> DataPlatform
    ERP[Sistema ERP] --> DataPlatform
    DataPlatform --> BI[Herramienta BI]
```
**Audiencia:** Ejecutivos, negocio.

### Nivel 2: Contenedores (Containers)
Descompone en aplicaciones/servicios ejecutables.
```mermaid
graph LR
    Ingestion[Servicio Ingesta] --> Lake[Data Lake S3]
    Lake --> Spark[Spark ETL]
    Spark --> DW[Data Warehouse]
    DW --> API[API REST]
    API --> Dashboard[Dashboard BI]
```
**Audiencia:** Arquitectos, tech leads.

### Nivel 3: Componentes (Components)
Detalle interno de un contenedor (clases, módulos).
- Menos común en arquitectura de datos, más en software.

### Nivel 4: Código (Code)
Diagramas UML de clases (raramente usado en data, más en ingeniería de software).

## 2. Flujos de Datos End-to-End con Swimlanes
**Swimlanes** separan responsabilidades por actor/sistema.

### Ejemplo: Pipeline de Órdenes
```mermaid
sequenceDiagram
    participant CRM
    participant IngestionAPI
    participant Bronze
    participant Silver
    participant Gold
    participant BI
    
    CRM->>IngestionAPI: POST /orders (JSON)
    IngestionAPI->>Bronze: Write raw (Parquet)
    Bronze->>Silver: ETL: validate, deduplicate
    Silver->>Gold: Aggregate by date/region
    Gold->>BI: Query: daily revenue
```

**Beneficios:**
- Traza responsabilidad por capa.
- Identifica latencias entre pasos.
- Base para SLAs (cada paso tiene tiempo esperado).
