## 2. Métricas de Alineación
| Métrica | Descripción | Fórmula | Objetivo |
|---------|-------------|---------|----------|
| Domain Coverage | % KPIs con entidades semánticas mapeadas | KPIs_mapeados / KPIs_totales | > 85% |
| Redundancia Semántica | Conceptos duplicados en diferentes contextos | conceptos_duplicados / conceptos_totales | < 5% |
| Time-to-KPI | Tiempo en incorporar nuevo KPI (idea→producción) | días | < 30 días |
| Data Product Alignment | % data products vinculados a objetivo estratégico | productos_alineados / total_productos | > 70% |
| Ontology Evolution Velocity | Versiones/semestre vs backlog | versiones_semestre | >= 2 |

`````python
import random
import statistics
from dataclasses import dataclass

random.seed(42)

@dataclass
class AlignmentMetrics:
    kpis_total: int
    kpis_mapped: int
    concepts_total: int
    concepts_duplicated: int
    kpi_delivery_days: list
    data_products_total: int
    data_products_aligned: int
    ontology_versions_semester: int

    def domain_coverage(self):
        return self.kpis_mapped / self.kpis_total * 100

    def redundancy(self):
        return self.concepts_duplicated / self.concepts_total * 100

    def time_to_kpi_avg(self):
        return statistics.mean(self.kpi_delivery_days)

    def data_product_alignment(self):
        return self.data_products_aligned / self.data_products_total * 100

metrics = AlignmentMetrics(
    kpis_total=40,
    kpis_mapped=34,
    concepts_total=180,
    concepts_duplicated=6,
    kpi_delivery_days=[random.randint(15, 55) for _ in range(12)],
    data_products_total=28,
    data_products_aligned=21,
    ontology_versions_semester=2
)
print('Cobertura de KPIs:', f'{metrics.domain_coverage():.1f}%')
print('Redundancia semántica:', f'{metrics.redundancy():.1f}%')
print('Tiempo promedio a KPI:', f'{metrics.time_to_kpi_avg():.1f} días')
print('Alineación de data products:', f'{metrics.data_product_alignment():.1f}%')
`````
