````python
from dataclasses import dataclass
from typing import Dict

@dataclass
class Readiness:
    ownership: int
    platform: int
    governance: int
    semantics: int
    security: int
    observability: int

    def score(self) -> float:
        vals = [self.ownership, self.platform, self.governance, self.semantics, self.security, self.observability]
        assert all(0 <= v <= 5 for v in vals), 'Valores deben estar en [0,5]'
        return sum(vals)/len(vals)

    def gaps(self) -> Dict[str,int]:
        return {k:v for k,v in self.__dict__.items() if isinstance(v,int) and v < 3}

# Ejemplo de evaluación inicial
current = Readiness(ownership=2, platform=3, governance=2, semantics=2, security=3, observability=2)
print('Score actual:', current.score())
print('Brechas:', current.gaps())
````

## 5. Matriz de Decisiones: Mesh vs Fabric vs Híbrido
Criterios: autonomía de dominios, complejidad técnica, cumplimiento, costos, time-to-value.

````python
import pandas as pd
criteria = ['autonomia','cumplimiento','costos','time_to_value','complejidad']
weights = {'autonomia':0.25,'cumplimiento':0.2,'costos':0.2,'time_to_value':0.2,'complejidad':0.15}
scores = pd.DataFrame({
    'criterio': criteria,
    'mesh':[5,4,3,4,3],
    'fabric':[3,4,4,3,4],
    'hibrido':[4,4,4,4,3]
})
def weighted(scores_row):
    return sum(scores_row[c]*weights[scores_row['criterio']] for c in ['mesh','fabric','hibrido'])
scores
````

````python
# Cálculo de totales ponderados
totals = {k:0.0 for k in ['mesh','fabric','hibrido']}
for _,row in scores.iterrows():
    for k in totals:
        totals[k] += row[k]*weights[row['criterio']]
totals
````
