# Taller: Arquitectura IoT Industrial en Tiempo Real (Edge + Streaming + IA)

Objetivo: diseñar una arquitectura edge-cloud para sensores industriales con ingestión en streaming, procesamiento de eventos, inferencia en tiempo real y feedback operativo.

---
## 1. Requisitos y Contexto
- Dispositivos: 100k sensores (vibración/temperatura)
- Latencia objetivo alerta: < 2 s
- Disponibilidad: 99.9%
- Cumplimiento: seguridad/privacidad de telemetría y auditoría de acciones

---
## 2. Arquitectura de Referencia (texto)
Edge Gateway (buffer + compresión + anonimización) → Event Broker (Kafka/Pulsar) → Stream Processing (Flink/Spark) → Feature Store Online → Inference Service (REST/gRPC) → Actuación/Alertas + Data Lake para histórico.

---
## 3. Latency Budget y Dimensionamiento Inicial

```python
# Cálculo simple de budget de latencia por etapa
stages = {
        'edge_buffer_ms': 50, 'network_ms': 150, 'broker_ms': 100, 'processing_ms': 300, 'inference_ms': 200, 'sink_ms': 50
      }
total = sum(stages.values())
target = 2000  # 2s
{'total_latency_ms': total, 'target_ms': target, 'headroom_ms': target-total}
```

---
## 4. Particionamiento y Throughput por Topic

```python
# Estimación de particiones para un throughput objetivo
events_per_sec = 50000  # 50k eps
per_partition_capacity = 2000  # eps por partición asumido
partitions = int((events_per_sec + per_partition_capacity - 1)/per_partition_capacity)
{'events_per_sec': events_per_sec, 'per_partition_capacity': per_partition_capacity, 'estimated_partitions': partitions}
```
